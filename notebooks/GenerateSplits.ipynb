```python
import json
import pandas as pd
from sklearn.model_selection import StratifiedKFold, train_test_split

DATASET_PATH = "../data/preprocesados/merge_clean.csv"
OUT_PATH = "../data/preprocesados/cv_splits_10.json"
```

```python
df = pd.read_csv(DATASET_PATH)

rids = df["RID"].values
diags = df["DIAGNOSIS_T2"].values

skf = StratifiedKFold(n_splits=10, shuffle=True, random_state=42)
splits = {}

for i, (train_val, test) in enumerate(skf.split(rids, diags)):
    train_val_rids = rids[train_val]
    test_rids = rids[test]

    train_rids, val_rids = train_test_split(train_val_rids, test_size=1/9, stratify=diags[train_val], random_state=42)[:2]

    splits[f"fold_{i}"] = {
        "train_rids": [int(x) for x in train_rids],
        "val_rids": [int(x) for x in val_rids],
        "test_rids": [int(x) for x in test_rids],
    }

with open(OUT_PATH, "w") as f:
    json.dump(splits, f, indent=4)
```
